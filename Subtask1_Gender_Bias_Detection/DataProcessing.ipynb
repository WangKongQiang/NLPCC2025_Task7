# train

```python
import json
with open('./NLPCC-2025-Shared-Task-7/data/train/biased.json', 'r', encoding='utf-8') as file:
    biased_data = json.load(file)
sentence_list=[]
labels_list=[]
for i_data in biased_data:
    sentence_list.append(i_data['ori_sentence'])
    labels_list.append(i_data['bias_labels'])
with open('./NLPCC-2025-Shared-Task-7/data/train/non-biased.json', 'r', encoding='utf-8') as file:
    non_biased_data = json.load(file)
for i_data in non_biased_data:
    sentence_list.append(i_data['text'])
    labels_list.append([0,0,0])
```

```python
import pandas as pd
data = {
        'ori_sentence': sentence_list,
        'bias_labels': labels_list
    }
df = pd.DataFrame(data)
df.to_csv(r"train_labels.csv", index=False)
```
